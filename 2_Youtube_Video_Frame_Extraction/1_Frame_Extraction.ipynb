```python
import cv2
import os
from tqdm.notebook import tqdm

# Define your directories
input_dir = "../1_Youtube_Downloaded_Video/"
output_base_dir = "Extracted_Videos/"

# Ensure the base output directory exists
os.makedirs(output_base_dir, exist_ok=True)

# Get a sorted list of video files in the input directory
# This filters out hidden files (like .DS_Store) and grabs valid files
valid_extensions = ('.mp4', '.mkv', '.avi', '.mov', '.webm')
video_files = [f for f in os.listdir(input_dir) if f.lower().endswith(valid_extensions)]
video_files.sort() # Sorting ensures consistent Video_1, Video_2 numbering

if not video_files:
    print(f"No video files found in {input_dir}")
else:
    print(f"Found {len(video_files)} videos. Starting extraction...")

for i, video_file in enumerate(video_files, start=1):
    video_path = os.path.join(input_dir, video_file)
    
    # Create the specific output directory for this video (e.g., Extracted_Videos/Video_1)
    video_out_dir = os.path.join(output_base_dir, f"Video_{i}")
    os.makedirs(video_out_dir, exist_ok=True)
    
    print(f"\nProcessing Video {i}/{len(video_files)}: {video_file}")
    
    # Open the video stream
    cap = cv2.VideoCapture(video_path)
    
    if not cap.isOpened():
        print(f"Error: Could not open {video_path}")
        continue
        
    # Get total frames for the progress bar
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    frame_count = 0
    saved_count = 0
    
    # Use tqdm to show a progress bar in Jupyter
    with tqdm(total=total_frames, desc=f"Video {i} Progress") as pbar:
        while cap.isOpened():
            ret, frame = cap.read()
            
            # If ret is False, we've reached the end of the video
            if not ret:
                break 
            
            # Save every 3rd frame (0, 3, 6, 9...)
            if frame_count % 240 == 0:
                # Format filename with zero-padding (e.g., frame_000001.jpg)
                out_filename = os.path.join(video_out_dir, f"frame_{saved_count:06d}.jpg")
                cv2.imwrite(out_filename, frame)
                saved_count += 1
                
            frame_count += 1
            pbar.update(1) # Update progress bar
            
    cap.release()
    print(f"Finished Video {i}: Saved {saved_count} frames to {video_out_dir}/")

print("\nAll videos processed successfully!")
```
