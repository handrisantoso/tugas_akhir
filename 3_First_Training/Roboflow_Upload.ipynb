```python
from roboflow import Roboflow

rf = Roboflow(api_key="")
project = rf.workspace("final-project-otter").project("billiards-zlynh")

# Get a list of all versions in this project
versions = project.versions()

if not versions:
    print("Error: No versions found! Please go to Roboflow and click 'Generate' first.")
else:
    # Grab the most recent version automatically
    latest_version = versions[-1]
    print(f"Found version {latest_version.version}. Uploading weights...")
    
    # Deploy to that specific version
    latest_version.deploy(
        model_type="yolov11", 
        model_path="Pool_Detection/yolo11n_run/weights/" 
    )
    print("Upload complete!")
```
