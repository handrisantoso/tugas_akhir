# Alarm Sound Test

Run the cells below to preview different alarm sounds. The continuous alarm cell keeps playing until you stop it with `Ctrl+C` in the terminal or interrupt the notebook kernel.

```python
import time
import winsound

ALARM_SOUNDS = {
    "classic": [(1200, 700), (0, 200)],
    "urgent": [(1800, 180), (0, 80), (1800, 180), (0, 80), (1400, 350), (0, 200)],
    "siren": [(700, 180), (900, 180), (1100, 180), (1300, 180), (1100, 180), (900, 180)],
    "high_low": [(1500, 300), (900, 300), (1500, 300), (900, 300)],
    "soft_chime": [(880, 250), (1175, 250), (1320, 450), (0, 250)],
}

def play_alarm(name, repeats=1):
    pattern = ALARM_SOUNDS[name]
    for _ in range(repeats):
        for frequency, duration_ms in pattern:
            if frequency == 0:
                time.sleep(duration_ms / 1000)
            else:
                winsound.Beep(frequency, duration_ms)

print("Available alarm sounds:")
for sound_name in ALARM_SOUNDS:
    print(f"- {sound_name}")
```

## Preview One Sound

Change `selected_sound` to one of: `classic`, `urgent`, `siren`, `high_low`, or `soft_chime`.

```python
selected_sound = "soft_chime"

play_alarm(selected_sound, repeats=3)
```

## Preview All Sounds

Run this cell to hear every alarm option one by one.

```python
for sound_name in ALARM_SOUNDS:
    print(f"Playing: {sound_name}")
    play_alarm(sound_name, repeats=2)
    time.sleep(0.6)
```

## Continuous Alarm

Run this cell to test the same behavior used after training. It keeps playing until you stop it with `Ctrl+C` in the terminal or interrupt the notebook kernel.

```python
selected_sound = "urgent"

print("Alarm is running. Press Ctrl+C or interrupt the kernel to stop it.")
try:
    while True:
        play_alarm(selected_sound, repeats=1)
        time.sleep(0.2)
except KeyboardInterrupt:
    print("Alarm stopped.")
```
