```python
import yaml
from ultralytics import YOLO
import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'TRUE'

# 1. Create the dataset.yaml file required by YOLO
yaml_content = {
    'path': os.path.abspath('../Images/YOLO_Dataset'),
    'train': 'images/train',
    'val': 'images/val',
    'test': 'images/test',
    'names': {
        0: '1-ball',
        1: '2-ball',
        2: '3-ball',
        3: '4-ball',
        4: '5-ball',
        5: '6-ball',
        6: '7-ball',
        7: '8-ball',
        8: '9-ball',
        9: 'cue',
        10: 'cue_stick',
        11: 'pocket',
        12: 'rack'
    }
}

with open('dataset.yaml', 'w') as f:
    yaml.dump(yaml_content, f, sort_keys=False)

print("dataset.yaml created successfully.")

# 2. Download YOLOv8n and Train
model = YOLO('yolov8n.pt') 

print("Starting training...")
results = model.train(
    data='dataset.yaml',
    epochs=100,
    imgsz=640,
    batch=8,
    patience=10,
    workers=4,
    close_mosaic=20,
    cos_lr=True,
    project=os.path.abspath('../Trained_Models'), 
    name='YOLOv8n'
)

# 3. Start an alarm after training finishes. Press Ctrl+C to stop it.
import time
import winsound

print("Training finished. Alarm is running. Press Ctrl+C to stop it.")
try:
    while True:
        winsound.Beep(1200, 700)
        time.sleep(0.2)
except KeyboardInterrupt:
    print("Alarm stopped.")
```
