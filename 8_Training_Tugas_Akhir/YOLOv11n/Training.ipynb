```python
import yaml
from ultralytics import YOLO
import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'TRUE'

no_aug_settings = {
    'hsv_h': 0.0,
    'hsv_s': 0.0,
    'hsv_v': 0.0,
    'degrees': 0.0,
    'translate': 0.0,
    'scale': 0.0,
    'shear': 0.0,
    'perspective': 0.0,
    'flipud': 0.0,
    'fliplr': 0.0,
    'bgr': 0.0,
    'mosaic': 0.0,
    'mixup': 0.0,
    'copy_paste': 0.0,
    'close_mosaic': 0,
}

# 1. Create the dataset.yaml file required by YOLO
yaml_content = {
    'path': os.path.abspath('../Images/YOLO_Dataset'),
    'train': 'images/train',
    'val': 'images/val',
    'test': 'images/test',
    'augmentations': no_aug_settings,
    'names': {
        0: '1-ball',
        1: '2-ball',
        2: '3-ball',
        3: '4-ball',
        4: '5-ball',
        5: '6-ball',
        6: '7-ball',
        7: '8-ball',
        8: '9-ball',
        9: 'cue',
        10: 'cue_stick',
        11: 'pocket',
        12: 'rack'
    }
}

with open('dataset.yaml', 'w') as f:
    yaml.dump(yaml_content, f, sort_keys=False)

print("dataset.yaml created successfully.")

# 2. Download YOLOv11n and Train
model = YOLO('yolo11n.pt') 

print("Starting training...")
results = model.train(
    data='dataset.yaml',
    epochs=100,
    imgsz=640,
    batch=8,
    patience=10,
    workers=4,
    cos_lr=True,
    project=os.path.abspath('../Trained_Models'), 
    name='ImageFixYOLO11n',
    **no_aug_settings
)
```
