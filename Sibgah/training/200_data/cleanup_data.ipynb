```python
import os
import shutil
import pandas as pd

DATASET_DIR = "../dataset"
OUTPUT_DIR = "cleaned_data"
LABELS = ["down", "idle", "left", "right", "up"]
MIN_ROWS = 10

stats = {"kept": 0, "removed": 0}

for label in LABELS:
    input_folder = os.path.join(DATASET_DIR, label)
    output_folder = os.path.join(OUTPUT_DIR, label)
    os.makedirs(output_folder, exist_ok=True)

    csv_files = [f for f in os.listdir(input_folder) if f.endswith(".csv")]

    for filename in csv_files:
        src = os.path.join(input_folder, filename)
        df = pd.read_csv(src)

        if len(df) >= MIN_ROWS:
            shutil.copy(src, os.path.join(output_folder, filename))
            stats["kept"] += 1
        else:
            stats["removed"] += 1
            print(f"[REMOVED] {label}/{filename} — only {len(df)} rows")

print(f"\nDone. Kept: {stats['kept']}, Removed: {stats['removed']}")
```
