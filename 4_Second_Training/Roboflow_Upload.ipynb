```python
from roboflow import Roboflow

# Initialize Roboflow with your private API key
rf = Roboflow(api_key="")

# Connect to your workspace
workspace = rf.workspace("final-project-otter")

print("Uploading model directly to workspace...")

# Deploy the model using the workspace method
workspace.deploy_model(
    model_type="yolov11n",                     # Updated to match the model you trained
    model_path="Pool_Detection/yolo11n_run",  # Path to the directory containing the 'weights' folder
    project_ids=["billiards-zlynh"],          # List containing your target project ID
    model_name="pool-ball-detector-v1",       # Valid name (letters, numbers, dashes)
    filename="weights/best.pt"                # Path to the specific weights file inside model_path
)

print("Model successfully deployed to your Roboflow workspace!")
```
